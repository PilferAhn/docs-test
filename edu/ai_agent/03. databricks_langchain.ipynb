## databricks-langchain
Databricks는 LangChain과의 통합을 위해 전용 라이브러리인 databricks-langchain을 제공합니다. <br>이를 통해 Databricks의 Model Serving 엔드포인트에 배포된 모델(LLM)을 LangChain의 표준 인터페이스로 쉽게 연결할 수 있습니다.

문서 : https://docs.databricks.com/aws/en/large-language-models/langchain

```python
%pip install databricks_langchain
dbutils.library.restartPython()
```

```python
from databricks_langchain import ChatDatabricks

llm = ChatDatabricks(
    #Databricks 'Serving' 메뉴에 생성된 모델 엔드포인트의 이름
    endpoint="databricks-gpt-5-2",
    temperature=0 
)
```

```python
query ="오늘 저녁은 뭘 먹을까 ?" #원하는 질의를 넣어주세요

result = llm.invoke(query)
```
